```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
wine = load_wine(as_frame=True)

df = wine.frame

df.head()
```

```python
print("Dataset shape:", df.shape)

print("Columns:")

print(df.columns.tolist())
df.info()
```

```python
df.to_csv("wine_dataset.csv", index=False)
```

```python
print("Mean:", df['alcohol'].mean())
print("Median:", df['alcohol'].median())
print("Minimum:", df['alcohol'].min())
print("Maximum:", df['alcohol'].max())
print("Standard deviation:", df['alcohol'].std())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['alcohol'], bins=15, kde=True)

plt.title("Distribution of alcohol")
plt.xlabel("alcohol")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 3))

sns.boxplot(x=df['alcohol'])
plt.title("boxplot of alcohol")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x='alcohol',
    y='malic_acid',
    hue='target',
    palette='viridis',
    s=80
)
plt.title('Alcohol vs Malic_acid')
plt.xlabel('Alcohol')
plt.ylabel('Malic_acid')
plt.legend(title='Wine Class')
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x='flavanoids',
    y='hue',
    hue='target',
    palette='viridis',
    s=80
)
plt.title('Flavaniods vs total phenols')
plt.xlabel('Flavanoids')
plt.ylabel('total phenol')
plt.legend(title='Wine Class')
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(
    data=df,
    x='target',
    y='alcohol',
    hue='target',
    palette='viridis'
)
plt.title('Alcohol distribution across wine classes')
plt.xlabel('Wine Class')
plt.ylabel('Alcohol')
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.boxplot(
    data=df,
    x='target',
    y='malic_acid',
    hue='target',
    palette='viridis'
)
plt.title('Malic acid distribution across wine classes')
plt.xlabel('Wine Class')
plt.ylabel('Malic acid')
plt.show()
```

```python
selected_features = [
    'alcohol',
    'malic_acid',
    'flavanoids',
    'color_intensity',
]
sns.pairplot(
    df[selected_features + ['target']],
    hue='target',
    palette='viridis',
    diag_kind='kde'
)
plt.show()
```

```python
correlation = X.corr()
correlation
```

```python
plt.figure(figsize=(14, 10))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    mask=np.triu(correlation)
)
plt.title('correlation of heatmap wine features')

plt.show
```
